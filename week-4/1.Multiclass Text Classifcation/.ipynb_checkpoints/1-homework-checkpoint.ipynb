## Generate TFIDF features

```python
#Bag of words - List of unique words in a text corpus
#The IDF value of a rare word is high, whereas the IDF value of a frequent word is low
```

```python
newsgroups_train = fetch_20newsgroups(subset='train', remove=('headers', 'footers', 'quotes'))
newsgroups_test = fetch_20newsgroups(subset='test', remove=('headers', 'footers', 'quotes'))
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
from time import time
```

```python
tfidf_vec = TfidfVectorizer(stop_words='english',max_features=5000)
```

```python
X_train = tfidf_vec.fit_transform(newsgroups_train.data)
X_test = tfidf_vec.transform(newsgroups_test.data)
```

```python
y_train = newsgroups_train.target
y_test = newsgroups_test.target
```

```python
print(X_train.shape)
print(X_test.shape)
```

## Train a Logistic Regression model off of said features

```python
model = LogisticRegression(max_iter=1000)
model = model.fit(X_train, y_train)
```

## Calculate a multiclass classification report

```python
y_pred = model.predict(X_test)
```

```python
# Quick view of metrics
print(classification_report(y_test, y_pred))
```

```python
# Generate classification report as dictionary
report = classification_report(y_test, y_pred, target_names=newsgroups_train.target_names, output_dict=True)
report
```

```python
from sklearn.multiclass import OneVsRestClassifier
from sklearn.svm import SVC
```

```python
model = OneVsRestClassifier(SVC())
model = model.fit(X_train, y_train)
```

```python
y_pred = model.predict(X_test)
```

```python
# Quick view of metrics
print(classification_report(y_test, y_pred))
```
